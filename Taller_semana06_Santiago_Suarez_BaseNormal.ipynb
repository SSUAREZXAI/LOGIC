### 6.2 Corrección de Código
Reescribe el fragmento de código de forma correcta en la celda de abajo y añade comentarios detallados explicando el impacto del ámbito (scope) y ciclo de vida de las variables en la memoria.

````python
# Datos de ejemplo para que la celda se pueda ejecutar
registros = [1500.0, 5000.0, 4200.0, 3900.0]
condicion = True

# Antes del ciclo, la variable 'contador' se crea en memoria una sola vez.
# Esta línea se ejecuta únicamente al inicio, antes de entrar al bucle.
# Mientras el ciclo da vueltas, la variable mantiene de forma acumulativa el total
# de registros que cumplen la condición, incrementándose paso a paso.
contador = 0

for registro in registros:
    if condicion:
        # En cada vuelta del ciclo, el valor anterior de 'contador' se conserva
        # y se le suma 1. No se reinicia ni se borra; el sistema toma el número
        # actual que está en memoria y lo actualiza con el nuevo total.
        contador += 1

# Al terminar el ciclo, 'contador' conserva el número total de coincidencias.
# Esto es posible gracias al "scope" (ámbito) en Python: los ciclos 'for' no
# aíslan sus variables, por lo que 'contador' vive fuera del bucle. Si se hubiera
# creado dentro del 'for', se habría reiniciado a 0 en cada vuelta, provocando
# un error donde el resultado final siempre sería 0 o 1, sin importar cuántos
# registros cumplieran la condición.
# Ciclo de vida: la variable se crea cuando se ejecuta contador = 0, cambia de valor
# (se actualiza) durante el ciclo, y sigue existiendo después de él hasta que el
# programa termina (o hasta que se reinicia la sesión de Colab).


print(contador)
````
