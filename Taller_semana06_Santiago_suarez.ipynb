## 7. Retos 10 y 11 — Integración y Resumen Final del Dataset

Llegó el momento de unificar toda la arquitectura de lógica computacional en un único ciclo estructurado de procesamiento para tu dataset real.

**Requisitos mínimos:**
*   Debe recorrer de principio a fin la variable `mi_dataset_lista`.
*   Si construiste funciones de clasificación o validación en las Semanas 4 o 5, **debes reutilizarlas** aquí mediante llamadas funcionales. ¡No dupliques código!

````python
def clasificar_registro(fila):
    condicion = fila[columna_condicion]
    if condicion in ("Conductor", "Peatón", "Pasajero"):
        return condicion
    return "Sin información"

total = 0
excluidas = 0
cont_hombre = 0
cont_mujer = 0
cont_conductor = 0
cont_peaton = 0
cont_pasajero = 0
cont_sin_info = 0
acum_casos = 0.0
hay_primera_infancia = False

for fila in mi_dataset_lista:
    if fila[columna_localidad] == "BOGOTA":
        excluidas += 1
        continue

    total += 1
    acum_casos += float(fila[columna_casos])

    if fila[columna_sexo] == "Hombre":
        cont_hombre += 1
    elif fila[columna_sexo] == "Mujer":
        cont_mujer += 1

    categoria = clasificar_registro(fila)
    if categoria == "Conductor":
        cont_conductor += 1
    elif categoria == "Peatón":
        cont_peaton += 1
    elif categoria == "Pasajero":
        cont_pasajero += 1
    else:
        cont_sin_info += 1
    if fila[columna_ciclo].startswith("(00 a 05)"):
        hay_primera_infancia = True
print(f"Total de registros (casos únicos): {total}")
print(f"Filas excluidas (LOCALIDAD = Bogotá): {excluidas}")
print(f"Hombres: {cont_hombre} | Mujeres: {cont_mujer}")
print(f"Conductores: {cont_conductor} | Peatones: {cont_peaton} | Pasajeros: {cont_pasajero} | Sin información: {cont_sin_info}")
print(f"Acumulado de casos: {acum_casos}")
print(f"¿Hay víctimas de Primera Infancia?: {hay_primera_infancia}")
assert cont_hombre + cont_mujer == total
assert cont_conductor + cont_peaton + cont_pasajero + cont_sin_info == total
assert acum_casos == total
print("Validaciones correctas: las categorías suman el total.")
````
